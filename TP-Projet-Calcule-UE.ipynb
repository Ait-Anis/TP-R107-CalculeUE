```python
#version 3
import random 
note1=[[(f"R{n}",random.randint(5,17)) for n in range(101,116)]+[(f"SAE{i}",random.randint(5,17)) for i in range(6)]]+[[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]]
note2=[[(f"R{n}",random.randint(5,17)) for n in range(101,116)]+[(f"SAE{i}",random.randint(5,17)) for i in range(6)]]+[[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]]
note3=[[(f"R{n}",random.randint(5,17)) for n in range(101,116)]+[(f"SAE{i}",random.randint(5,17)) for i in range(6)]]+[[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]]



def ue_dic(note):
    dico={}
    i=0
    for e in note[0]:   #on se balade dans les notes des sae et et des ressources
        dico[e[0]]=(e[1],note[1][i])    #on créé notre dico sous forme:( k : (v[0],v[1])   )
        i+=1
    return dico


        

def calcul_dic():
    ue1=ue_dic(note1)   #on créé nos dico d'ue a l'aide d'une variable global note1
    res_ue1=0
    ue2=ue_dic(note2)
    res_ue2=0
    ue3=ue_dic(note3)
    res_ue3=0
    i=0
    for k , v in ue1.items():   
        res_ue1+=v[0]*v[1]   #on multiplie la note par la valeur pour simplifier par la suite le calcule de la moyenne
        i+=v[1]        
    res_ue1= res_ue1/i      #on calcule notre moyenne
    i=0
    for k , v in ue2.items():
        res_ue2+=v[0]*v[1]
        i+=v[1]
    res_ue2= res_ue2/i
    i=0
    for k , v in ue3.items():
        res_ue3+=v[0]*v[1]
        i+=v[1]
    res_ue3= res_ue3/i

    return(res_ue1,res_ue2,res_ue3)

calcul_dic()
    
    

    
```

```python
import matplotlib.pyplot as plt
import numpy as np

note3=[random.randint(5,17) for n in range(21)]

def moy(coef):
    moy=round(np.average(note3,weights=coef),3)
    return moy


def annee():
    UE1=moy(coeff1)
    UE2=moy(coeff2)
    UE3=moy(coeff3)
    l_ue=[UE1,UE2,UE3]
    color_ue=[]
    y=20
    color=""
    for elt in l_ue:
        if elt>=10:
            color="g"
        elif 8<=elt<10:
            color="orange"
        elif 8>elt:
            color="r"
        color_ue.append(color)
    plt.bar(['UE1','UE2','UE3'],[UE1,UE2,UE3],color=color_ue,y=0)
    plt.errorbar(3,10,yerr=10,color='w')
    plt.show()

    

annee()
    
    
    


    
```
